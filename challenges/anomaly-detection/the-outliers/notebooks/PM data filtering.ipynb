```python
#importing all the needed libraries

from pathlib import Path
import tarfile
import pandas as pd
import numpy as np
import re
```

```python
# Configuration
input_dir = Path("/workspace/lustre/work/pdl18409/shared/The_outliers/processed_data_stefan/SPOKES events")
output_dir = Path("/workspace/lustre/work/pdl18409/shared/The_outliers/processed_data_stefan/SPOKES grouped by cavity name")

source_column = "pv"
group_column = "cavity_name"

output_dir.mkdir(parents=True, exist_ok=True)

# Store groups across all input files
groups = {}

for file in input_dir.glob("*.parquet"):
    print(f"Reading: {file.name}")

    df = pd.read_parquet(file)

    # Create the new column: keep everything before the second colon
    df[group_column] = (
        df[source_column]
        .str.split(":")
        .str[:2]
        .str.join(":")
    )

    # Group rows by the new column
    for value, group in df.groupby(group_column, dropna=False):
        key = "<MISSING>" if pd.isna(value) else str(value)

        if key not in groups:
            groups[key] = []

        groups[key].append(group)

# Save each group as a separate Parquet file
used_filenames = set()

for value, dataframes in groups.items():
    combined_df = pd.concat(dataframes, ignore_index=True)

    # Make the group value safe for use as a filename
    filename = re.sub(r'[<>:"/\\|?*\x00-\x1f]', "_", value).strip()

    if not filename or filename in {".", ".."}:
        filename = "empty_value"

    base_filename = filename
    i = 1

    while filename.lower() in used_filenames:
        filename = f"{base_filename}_{i}"
        i += 1

    used_filenames.add(filename.lower())

    output_file = output_dir / f"{filename}.parquet"
    combined_df.to_parquet(output_file, index=False)

    print(f"Saved {output_file.name}: {len(combined_df)} rows")

print(f"\nDone! Created {len(groups)} grouped files.")
```
