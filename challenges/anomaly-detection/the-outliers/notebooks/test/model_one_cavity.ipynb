```python
import math
import torch
import gc

# 前回確保したXがあればメモリを解放
if "X" in globals():
    del X
    gc.collect()

file_info = []

# 1回目：各ファイルの形状とcycle_idを調査
for path, digitizer in zip(files, digitizers):
    obj = torch.load(
        path,
        map_location="cpu",
        weights_only=False
    )

    data = obj["data"]
    cycle_id = obj["cycle_id"]

    if isinstance(cycle_id, torch.Tensor):
        cycle_ids = cycle_id.detach().cpu().tolist()
    else:
        cycle_ids = list(cycle_id)

    if data.shape[0] != len(cycle_ids):
        raise ValueError(
            f"{path.name}: "
            f"data={data.shape[0]}行、cycle_id={len(cycle_ids)}個"
        )

    if len(cycle_ids) != len(set(cycle_ids)):
        raise ValueError(
            f"{path.name}: cycle_idに重複があります"
        )

    feature_count = math.prod(data.shape[1:])

    file_info.append({
        "path": path,
        "digitizer": digitizer,
        "cycle_ids": cycle_ids,
        "data_shape": tuple(data.shape),
        "feature_count": feature_count,
        "dtype": data.dtype,
    })

    print(
        path.name,
        "shape =", tuple(data.shape),
        "1行にしたとき =", feature_count
    )

    del obj, data, cycle_id
    gc.collect()
```

```python
common_set = set(file_info[0]["cycle_ids"])

for info in file_info[1:]:
    common_set &= set(info["cycle_ids"])

# 最初のファイルに登場する順序を維持
common_cycle_ids = [
    cycle_id
    for cycle_id in file_info[0]["cycle_ids"]
    if cycle_id in common_set
]

if len(common_cycle_ids) == 0:
    raise ValueError(
        "4ファイルに共通するcycle_idがありません"
    )

print("共通cycle数:", len(common_cycle_ids))

for info in file_info:
    print(
        info["digitizer"],
        "元のcycle数:",
        len(info["cycle_ids"]),
        "使用するcycle数:",
        len(common_cycle_ids)
    )
```

```python
from pathlib import Path
import torch
import gc

root = Path(
    "/workspace/lustre/work/pdl18409/shared/data/anomaly-detection/"
    "teams/the-outliers/DD/step3_DD_tensors/"
    "HBL_DD_tensors"
)

cavity = "010RFC"
digitizers = [101, 201, 301, 401]

files = [
    root / f"tensor_HBL-{cavity}_RFS-DIG-{digitizer}.pt"
    for digitizer in digitizers
]

for path in files:
    print(path.name, path.exists())
```

```python
n_rows = len(common_cycle_ids)
total_features = sum(
    info["feature_count"]
    for info in file_info
)

X = torch.empty(
    (n_rows, total_features),
    dtype=file_info[0]["dtype"]
)

column_start = 0

for info in file_info:
    obj = torch.load(
        info["path"],
        map_location="cpu",
        weights_only=False
    )

    data = obj["data"]

    # cycle_idからファイル内の行番号を検索
    cycle_to_row = {
        cycle_id: row
        for row, cycle_id in enumerate(info["cycle_ids"])
    }

    row_indices = torch.tensor(
        [
            cycle_to_row[cycle_id]
            for cycle_id in common_cycle_ids
        ],
        dtype=torch.long
    )

    # 共通cycleだけを取り出して1行化
    flattened = data.index_select(
        0,
        row_indices
    ).reshape(n_rows, -1)

    column_end = column_start + flattened.shape[1]

    X[:, column_start:column_end].copy_(flattened)

    print(
        f"digitizer {info['digitizer']}:",
        tuple(data.shape),
        "→",
        tuple(flattened.shape),
        f"列 {column_start}:{column_end}"
    )

    column_start = column_end

    del obj, data, row_indices, flattened
    gc.collect()
```

```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader, TensorDataset
from sklearn.manifold import TSNE


n = len(X)

n_train = int(n * 0.8)
n_val = int(n * 0.1)

# ランダムにせず、順番通りに分割
train_data = X[:n_train]
val_data = X[n_train:n_train+n_val]
test_data = X[n_train+n_val:]

print("Train:", train_data.shape)
print("Validation:", val_data.shape)
print("Test:", test_data.shape)

# ========================================
# 3. 標準化
# ========================================

# Trainデータからチャンネルごとの平均と標準偏差を計算
mean_train = train_data.mean(dim=-1, keepdim=True)
std_train = train_data.std(dim=-1, keepdim=True)
mean_val = val_data.mean(dim=-1, keepdim=True)
std_val = val_data.std(dim=-1, keepdim=True)
mean_test = test_data.mean(dim=-1, keepdim=True)
std_test = test_data.std(dim=-1, keepdim=True)

# 同じ平均・標準偏差で標準化
train_data = (train_data - mean_train) / std_train
val_data = (val_data - mean_val) / std_val
test_data = (test_data - mean_test) / std_test

# ========================================
# 4. DataLoaderの作成
# ========================================

batch_size = 4

train_loader = DataLoader(
    TensorDataset(train_data),
    batch_size=batch_size,
    shuffle=False
)

val_loader = DataLoader(
    TensorDataset(val_data),
    batch_size=batch_size,
    shuffle=False
)

test_loader = DataLoader(
    TensorDataset(test_data),
    batch_size=batch_size,
    shuffle=False
)
```

```python
# ========================================
# 5. Autoencoderモデル
# ========================================

class RFAutoencoder(nn.Module):
    def __init__(self, latent_dim=2):
        super().__init__()

        self.encoder = nn.Sequential(
            nn.Conv1d(6, 12, kernel_size=8, stride=4, padding=2),
            nn.GELU(),

            nn.Conv1d(12, 24, kernel_size=8, stride=4, padding=2),
            nn.GELU(),

            nn.Flatten(),
            nn.Linear(52824, latent_dim)
        )

        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 52824),
            nn.Unflatten(1, (24, 2201)),

            nn.ConvTranspose1d(
                24, 12,
                kernel_size=8,
                stride=4,
                padding=2,
                output_padding=1
            ),
            nn.GELU(),

            nn.ConvTranspose1d(
                12, 6,
                kernel_size=8,
                stride=4,
                padding=2,
                output_padding=1
            )
        )

    def forward(self, x):
        z = self.encoder(x)
        reconstructed = self.decoder(z)
        return reconstructed


# ========================================
# 6. モデルとLoss関数の設定
# ========================================

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model = RFAutoencoder(latent_dim=2).to(device)

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

print("Device:", device)

# ========================================
# 7. 学習
# ========================================

epochs = 20

train_losses = []
val_losses = []

for epoch in range(epochs):

    # ----- Train -----
    model.train()
    train_loss = 0

    for (x,) in train_loader:
        x = x.to(device)

        optimizer.zero_grad()

        output = model(x)
        loss = criterion(output, x)

        loss.backward()
        optimizer.step()

        train_loss += loss.item() * x.size(0)

    train_loss /= len(train_data)
    train_losses.append(train_loss)

    # ----- Validation -----
    model.eval()
    val_loss = 0

    with torch.no_grad():
        for (x,) in val_loader:
            x = x.to(device)

            output = model(x)
            loss = criterion(output, x)

            val_loss += loss.item() * x.size(0)

    val_loss /= len(val_data)
    val_losses.append(val_loss)

    print(
        f"Epoch {epoch+1}/{epochs} "
        f"Train Loss: {train_loss:.6f} "
        f"Val Loss: {val_loss:.6f}"
    )

torch.save(model, "model-MBL-090RFC_RFS-EVR-101.pt")
# ========================================
# 8. Lossのグラフ
# ========================================

plt.figure(figsize=(8, 5))

plt.plot(train_losses, label="Train Loss")
plt.plot(val_losses, label="Validation Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss (MSE)")
plt.title("Training and Validation Loss")
plt.legend()
plt.grid(True)
plt.savefig("loss_function-MBL-090RFC_RFS-EVR-101.pdf",dpi=300)
plt.show()

# ========================================
# 9. 再構成誤差の計算
# ========================================

def calculate_errors(loader):
    model.eval()
    errors = []

    with torch.no_grad():
        for (x,) in loader:
            x = x.to(device)

            reconstructed = model(x)

            # 各イベントの再構成誤差（MSE）
            error = ((x - reconstructed) ** 2).mean(dim=(1, 2))

            errors.extend(error.cpu().numpy())

    return np.array(errors)


train_errors = calculate_errors(train_loader)
val_errors = calculate_errors(val_loader)
test_errors = calculate_errors(test_loader)

print("Train mean error:", train_errors.mean())
print("Validation mean error:", val_errors.mean())
print("Test mean error:", test_errors.mean())

# ========================================
# 10. 再構成誤差のグラフ
# ========================================

plt.figure(figsize=(10, 5))

plt.hist(
    train_errors,
    label="Train"
)

plt.hist(
    val_errors,
    label="Validation"
)

plt.hist(
    test_errors,
    label="Test"
)

plt.xlabel("Reconstruction Error (MSE)")
plt.ylabel("Count")
plt.title("Reconstruction Error histogram (MSE)")
plt.legend()
plt.savefig("reconstruction_error_histogram-MBL-090RFC_RFS-EVR-101.pdf", dpi=300)
plt.show()
```
